```
import zipfile
import os

RAW_DIR = "/content/drive/MyDrive/Projek/Air Quality Monitoring in European Cities/data/raw"

for file in os.listdir(RAW_DIR):

    if file.endswith(".zip"):

        zip_path = os.path.join(RAW_DIR, file)

        print(f"Extracting {file}...")

        with zipfile.ZipFile(zip_path, 'r') as zip_ref:
            zip_ref.extractall(RAW_DIR)

print("Extraction selesai.")
```

```
import pandas as pd

RAW_DIR = "/content/drive/MyDrive/Projek/Air Quality Monitoring in European Cities/data/raw"

ancona = pd.read_csv(
    f"{RAW_DIR}/ancona_data.csv"
)

athens = pd.read_csv(
    f"{RAW_DIR}/athens_data.csv"
)

zaragoza = pd.read_csv(
    f"{RAW_DIR}/zaragoza_data.csv"
)
```

```
datasets = {
    "Ancona": ancona,
    "Athens": athens,
    "Zaragoza": zaragoza
}

for name, df in datasets.items():

    print("="*60)
    print(name)
    print("="*60)

    print(f"Rows    : {df.shape[0]:,}")
    print(f"Columns : {df.shape[1]}")
    print()
```
